```python
from medpy.io import load
import matplotlib.pyplot as plt
import numpy as np
import plotly.graph_objects as go
from scipy import ndimage
```

```python
mha_file_path = "Normal001-MRA.mha"
image_data, image_header = load(mha_file_path)

print(f"Image data shape: {image_data.shape}")
print(f"Image header: {image_header}")
```

```python
# Get the middle slice for visualization
middle_slice = image_data.shape[2] // 2
slice_data = image_data[:, :, middle_slice]

plt.figure(figsize=(10, 10))
plt.imshow(slice_data, cmap='gray')
plt.title(f'MRA Image - Slice {middle_slice}')
plt.colorbar(label='Intensity')
plt.axis('off')

output_filename = 'mra_visualization.png'
plt.savefig(output_filename, dpi=150, bbox_inches='tight')
print(f"Visualization saved to {output_filename}")

plt.show()
```

```python
# Create a comprehensive 3D visualization with multiple slices
fig = plt.figure(figsize=(15, 12))

# Axial slices (view from top)
ax1 = plt.subplot(3, 3, 1)
slice_idx = image_data.shape[2] // 4
plt.imshow(image_data[:, :, slice_idx], cmap='gray')
plt.title(f'Axial - Slice {slice_idx}/{image_data.shape[2]}')
plt.axis('off')

ax2 = plt.subplot(3, 3, 2)
slice_idx = image_data.shape[2] // 2
plt.imshow(image_data[:, :, slice_idx], cmap='gray')
plt.title(f'Axial - Slice {slice_idx}/{image_data.shape[2]}')
plt.axis('off')

ax3 = plt.subplot(3, 3, 3)
slice_idx = 3 * image_data.shape[2] // 4
plt.imshow(image_data[:, :, slice_idx], cmap='gray')
plt.title(f'Axial - Slice {slice_idx}/{image_data.shape[2]}')
plt.axis('off')

# Sagittal slices (view from side)
ax4 = plt.subplot(3, 3, 4)
slice_idx = image_data.shape[0] // 4
plt.imshow(image_data[slice_idx, :, :].T, cmap='gray')
plt.title(f'Sagittal - Slice {slice_idx}/{image_data.shape[0]}')
plt.axis('off')

ax5 = plt.subplot(3, 3, 5)
slice_idx = image_data.shape[0] // 2
plt.imshow(image_data[slice_idx, :, :].T, cmap='gray')
plt.title(f'Sagittal - Slice {slice_idx}/{image_data.shape[0]}')
plt.axis('off')

ax6 = plt.subplot(3, 3, 6)
slice_idx = 3 * image_data.shape[0] // 4
plt.imshow(image_data[slice_idx, :, :].T, cmap='gray')
plt.title(f'Sagittal - Slice {slice_idx}/{image_data.shape[0]}')
plt.axis('off')

# Coronal slices (view from front)
ax7 = plt.subplot(3, 3, 7)
slice_idx = image_data.shape[1] // 4
plt.imshow(image_data[:, slice_idx, :].T, cmap='gray')
plt.title(f'Coronal - Slice {slice_idx}/{image_data.shape[1]}')
plt.axis('off')

ax8 = plt.subplot(3, 3, 8)
slice_idx = image_data.shape[1] // 2
plt.imshow(image_data[:, slice_idx, :].T, cmap='gray')
plt.title(f'Coronal - Slice {slice_idx}/{image_data.shape[1]}')
plt.axis('off')

ax9 = plt.subplot(3, 3, 9)
slice_idx = 3 * image_data.shape[1] // 4
plt.imshow(image_data[:, slice_idx, :].T, cmap='gray')
plt.title(f'Coronal - Slice {slice_idx}/{image_data.shape[1]}')
plt.axis('off')

plt.suptitle('3D MRA Visualization - Multiple Orientations', fontsize=16, y=0.98)
plt.tight_layout()

# Save the visualization
output_filename = 'mra_3d_visualization.png'
plt.savefig(output_filename, dpi=150, bbox_inches='tight')
print(f"3D visualization saved to {output_filename}")

plt.show()
plt.close()
print("Static visualization complete!")
```
