```python
# Function to plot Gromov-Wasserstein distances
def plot_gromov_wasserstein_distances(graphs_er, graphs_sbm):
  # Compute pairwise GW distances within ER dataset
  gw_distances_er = []
  for i, (er_graph1, _) in enumerate(graphs_er):
    er_matrix1 = compute_distance_matrix(er_graph1)
    for j, (er_graph2, _) in enumerate(graphs_er):
      if i < j:  # Avoid duplicate and self comparisons
        er_matrix2 = compute_distance_matrix(er_graph2)
        gw_transport, log_h = ot.gromov.gromov_wasserstein(
            er_matrix1, er_matrix2, loss_fun='square_loss', log=True)
        gw_distances_er.append(log_h['gw_dist'])
  # Compute pairwise GW distances within SBM dataset
  gw_distances_sbm = []
  for i, (sbm_graph1, _) in enumerate(graphs_sbm):
    sbm_matrix1 = compute_distance_matrix(sbm_graph1)
    for j, (sbm_graph2, _) in enumerate(graphs_sbm):
      if i < j:  # Avoid duplicate and self comparisons
        sbm_matrix2 = compute_distance_matrix(sbm_graph2)
        gw_transport, log_h = ot.gromov.gromov_wasserstein(
            sbm_matrix1, sbm_matrix2, loss_fun='square_loss', log=True)
        gw_distances_sbm.append(log_h['gw_dist'])
  # Plot GW distances for ER graphs
  plt.figure(figsize=(10, 5))
  plt.hist(gw_distances_er,
           bins=30,
           alpha=0.5,
           color='blue',
           label='GW Distances ER-ER')
  plt.xlabel('Gromov-Wasserstein Distance')
  plt.ylabel('Frequency')
  plt.title('Gromov-Wasserstein Distances Among ER Graphs')
  plt.legend()
  plt.show()
  # Plot GW distances for SBM graphs
  plt.figure(figsize=(10, 5))
  plt.hist(gw_distances_sbm,
           bins=30,
           alpha=0.5,
           color='green',
           label='GW Distances SBM-SBM')
  plt.xlabel('Gromov-Wasserstein Distance')
  plt.ylabel('Frequency')
  plt.title('Gromov-Wasserstein Distances Among SBM Graphs')
  plt.legend()
  plt.show()
```

```python
plot_gromov_wasserstein_distances(graphs_er5, graphs_sbm5)
```

```python
plot_gromov_wasserstein_distances(graphs_er, graphs_sbm)
```

```python
# Function to compute histogram of entries of pairwise distance matrices for the two graph datasets
def plot_distance_matrix_histograms(er_distance_matrices, sbm_distance_matrices):
    """Plots side-by-side histograms of entries in pairwise distance matrices for ER and SBM graphs.
    Args:
        er_distance_matrices: List of numpy arrays representing pairwise distances for ER graphs.
        sbm_distance_matrices: List of numpy arrays representing pairwise distances for SBM graphs.
    """
    # Flatten matrices into arrays of distances
    er_distances = np.hstack([matrix.flatten() for matrix in er_distance_matrices])
    sbm_distances = np.hstack([matrix.flatten() for matrix in sbm_distance_matrices])
    
    # Calculate the number of bins and bin edges
    bins = np.histogram_bin_edges(np.concatenate((er_distances, sbm_distances)), bins=30)
    
    # Plot histograms side-by-side
    width = (bins[1] - bins[0]) / 3  # width of each bar adjusting for overlap
    
    plt.figure(figsize=(12, 6))
    
    plt.hist(er_distances, bins=bins, alpha=0.5, color='blue', label='ER Graph Distances', 
             width=width, align='right', edgecolor='black')
    plt.hist(sbm_distances, bins=bins, alpha=0.5, color='green', label='SBM Graph Distances', 
             width=-width, align='left', edgecolor='black')
    
    plt.xlabel('Pairwise Distance')
    plt.ylabel('Frequency')
    plt.title('Histogram of Pairwise Distances in ER and SBM Graphs')
    plt.legend()
    plt.show()
```

```python
# prep the graphs by converting to pwise dist matrices
dist_er = [
    compute_distance_matrix(graph) for graph, label in graphs_er
]
dist_sbm = [
    compute_distance_matrix(graph) for graph, label in graphs_sbm
]
dist_er_5 = [
    compute_distance_matrix(graph) for graph, label in graphs_er5
]
dist_sbm_5 = [
    compute_distance_matrix(graph) for graph, label in graphs_sbm5
]
```

```python
# Results for experiments with ER EAP = 0.1, SBM EAP of 0.15 within communities and EAP of 0.05 between communities
plot_distance_matrix_histograms(dist_er, dist_sbm)
```

```python
# Results for experiments with ER EAP = 0.25, SBM EAP of 0.45 within communities and EAP of 0.05 between communities
plot_distance_matrix_histograms(dist_er_5, dist_sbm_5)
```

```python
### EXPERIMENT 6: Use adjacency matrices instead of pairwise distances.
def compute_adjacency_matrix(graph):
    """
    Computes the adjacency matrix for a given graph.
    
    Args:
        graph: A NetworkX graph.
    
    Returns:
        A numpy array representing the adjacency matrix where the value 
        at (i, j) denotes the presence of an edge between node i and node j.
    """
    # Initialize a matrix of zeros
    num_nodes = len(graph.nodes)
    adjacency_matrix = np.zeros((num_nodes, num_nodes))
    
    # Populate the adjacency matrix
    for i, node_i in enumerate(graph.nodes()):
        for j, node_j in enumerate(graph.nodes()):
            if graph.has_edge(node_i, node_j):
                adjacency_matrix[i, j] = 1.0
    return adjacency_matrix


prob_matrix_adj = [
    [0.15, 0.05],  # Probabilities related to block 1 connections
    [0.05, 0.15]  # Probabilities related to block 2 connections
]

graphs_er_adj = generate_erdos_renyi_graphs(100, 100, 0.1)
graphs_sbm_adj = generate_sbm_graphs(100, [50, 50], prob_matrix_adj)

full_graph_data_adj = graphs_er_adj + graphs_sbm_adj
adj_matrices = [
    compute_adjacency_matrix(graph) for graph, label in full_graph_data_adj
]
# Prepare labels
labels_adj = [label for graph, label in full_graph_data_adj]


prob_matrix_adj5 = [
    [0.45, 0.05],  # Probabilities related to block 1 connections
    [0.05, 0.45]  # Probabilities related to block 2 connections
]

graphs_er5_adj = generate_erdos_renyi_graphs(100, 100, 0.25)
graphs_sbm5_adj = generate_sbm_graphs(100, [50, 50], prob_matrix_adj5)

full_graph_data5_adj = graphs_er5_adj + graphs_sbm5_adj
adj_matrices5 = [
    compute_adjacency_matrix(graph) for graph, label in full_graph_data5_adj
]
# Prepare labels
labels5_adj = [label for graph, label in full_graph_data5_adj]
```

```python
exp4_adj = gw_barycenter_class_alg_var_nodes(adj_matrices,
                                      labels_adj,
                                      num_folds=10,
                                      num_nodes_negone=100,
                                      num_nodes_one=100,
                                      method="kfold")
```

```python
exp5_adj = gw_barycenter_class_alg_var_nodes(adj_matrices5,
                                      labels5_adj,
                                      num_folds=10,
                                      num_nodes_negone=100,
                                      num_nodes_one=100,
                                      method="kfold")
```
