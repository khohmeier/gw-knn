```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
```

```python
def create_and_save_graphs():
        # Create Erdos-Renyi graph
        n_nodes = 20
        p_er = 0.1
        er_graph = nx.erdos_renyi_graph(n_nodes, p_er, seed=42)

        # Create Stochastic Block Model graph
        n_communities = 2
        sizes = [10, 10]  # Two communities of 10 nodes each
        p_matrix = [
            [0.15, 0.05],  # Probability matrix for connections
            [0.05, 0.15]
        ]
        sbm_graph = nx.stochastic_block_model(sizes, p_matrix, seed=42)

        # Visualize Erdos-Renyi graph
        plt.figure(figsize=(8, 6))
        pos_er = nx.spring_layout(er_graph, seed=42)
        nx.draw(er_graph,
                pos_er,
                node_color='lightblue',
                with_labels=True,
                node_size=500,
                font_size=8)
        plt.title('Erdos-Renyi Graph (n=20, p=0.1)')
        plt.show()
        plt.savefig('erdos_renyi.png', bbox_inches='tight')
        plt.close()

        # Visualize SBM graph
        plt.figure(figsize=(8, 6))
        pos_sbm = nx.spring_layout(sbm_graph, seed=42, k=1)
        colors = ['lightblue'] * 10 + [
            'lightgreen'
        ] * 10  # Different colors for communities
        colors[4] = 'lightgreen'
        colors[13] = 'lightblue' 
        nx.draw(sbm_graph,
                pos_sbm,
                node_color=colors,
                with_labels=True,
                node_size=500,
                font_size=8)
        plt.title(
            'Stochastic Block Model Graph\n(n=20, p_within=0.15, p_between=0.05)')
        plt.show()
        plt.savefig('stochastic_block_model.png', bbox_inches='tight')
        plt.close()
```

```python
create_and_save_graphs()
```
